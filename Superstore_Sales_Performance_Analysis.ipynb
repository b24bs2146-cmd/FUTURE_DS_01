```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
df = pd.read_excel("/content/Sample - Superstore (1).xls")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
print("Number of rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset information:")
df.info()
```

```python
print("Missing values in each column:")
display(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nTotal duplicate rows:", df.duplicated().sum())

print("\nNegative Sales rows:", (df["Sales"] < 0).sum())

print("Negative Profit rows:", (df["Profit"] < 0).sum())

print("\nUnique categories:")
print(df["Category"].unique())

print("\nUnique regions:")
print(df["Region"].unique())
```

```python
print("Statistical Summary:")
display(df[["Sales", "Quantity", "Discount", "Profit"]].describe().round(2))

print("\nDate Range:")
print("First Order Date:", df["Order Date"].min())
print("Last Order Date:", df["Order Date"].max())

print("\nTotal Sales:", round(df["Sales"].sum(), 2))
print("Total Profit:", round(df["Profit"].sum(), 2))
print("Total Quantity Sold:", df["Quantity"].sum())
print("Unique Orders:", df["Order ID"].nunique())
print("Unique Customers:", df["Customer ID"].nunique())
```

```python
category_analysis = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values("Total_Sales", ascending=False)
    .round(2)
)
display(category_analysis)
```

```python
plt.figure(figsize=(8, 5))
plt.bar(
    category_analysis.index,
    category_analysis["Total_Sales"]
)
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales ($)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(
    category_analysis.index,
    category_analysis["Total_Profit"]
)
plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit ($)")
plt.tight_layout()
plt.show()
```

```python
df["Order Month"] = df["Order Date"].dt.to_period("M")

monthly_analysis = (
    df.groupby("Order Month")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .round(2)
)

display(monthly_analysis.head(12))
```

```python
plt.figure(figsize=(14, 5))
plt.plot(
    monthly_analysis.index.astype(str),
    monthly_analysis["Total_Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend (2014–2017)")
plt.xlabel("Month")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 5))
plt.plot(
    monthly_analysis.index.astype(str),
    monthly_analysis["Total_Profit"],
    marker="o"
)

plt.axhline(0, color="red", linewidth=1)
plt.title("Monthly Profit Trend (2014–2017)")
plt.xlabel("Month")
plt.ylabel("Total Profit ($)")
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
product_analysis = (
    df.groupby("Product Name")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values("Total_Sales", ascending=False)
)

top_10_products = product_analysis.head(10).round(2)

display(top_10_products)
```

```python
top_10_plot = top_10_products.sort_values("Total_Sales")

plt.figure(figsize=(12, 7))
plt.barh(
    top_10_plot.index,
    top_10_plot["Total_Sales"]
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales ($)")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

```python
region_analysis = (
    df.groupby("Region")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values("Total_Sales", ascending=False)
    .round(2)
)

display(region_analysis)
```

```python
plt.figure(figsize=(8, 5))
plt.bar(
    region_analysis.index,
    region_analysis["Total_Sales"]
)
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales ($)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(
    region_analysis.index,
    region_analysis["Total_Profit"]
)
plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit ($)")
plt.tight_layout()
plt.show()
```

```python
subcategory_analysis = (
    df.groupby("Sub-Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values("Total_Profit", ascending=True)
    .round(2)
)

display(subcategory_analysis)
```

```python
subcategory_plot = subcategory_analysis.sort_values("Total_Profit")

plt.figure(figsize=(10, 7))

plt.barh(
    subcategory_plot.index,
    subcategory_plot["Total_Profit"]
)

plt.axvline(0, color="black", linewidth=0.8)
plt.title("Total Profit by Sub-Category")
plt.xlabel("Total Profit ($)")
plt.ylabel("Sub-Category")

plt.tight_layout()
plt.show()
```

```python
df["Discount Band"] = pd.cut(
    df["Discount"],
    bins=[-0.001, 0, 0.10, 0.20, 0.30, 0.50, 1.0],
    labels=[
        "0%",
        "1-10%",
        "11-20%",
        "21-30%",
        "31-50%",
        "Above 50%"
    ]
)

discount_analysis = (
    df.groupby("Discount Band", observed=False)
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Average_Profit=("Profit", "mean"),
        Order_Rows=("Order ID", "count")
    )
    .round(2)
)

display(discount_analysis)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    discount_analysis.index.astype(str),
    discount_analysis["Total_Profit"]
)

plt.axhline(0, color="black", linewidth=0.8)
plt.title("Total Profit by Discount Band")
plt.xlabel("Discount Band")
plt.ylabel("Total Profit ($)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
best_sales_month = monthly_analysis["Total_Sales"].idxmax()
worst_sales_month = monthly_analysis["Total_Sales"].idxmin()

best_profit_month = monthly_analysis["Total_Profit"].idxmax()
worst_profit_month = monthly_analysis["Total_Profit"].idxmin()

print("Highest Sales Month:", best_sales_month)
print("Highest Monthly Sales ($):",
      round(monthly_analysis.loc[best_sales_month, "Total_Sales"], 2))

print("\nLowest Sales Month:", worst_sales_month)
print("Lowest Monthly Sales ($):",
      round(monthly_analysis.loc[worst_sales_month, "Total_Sales"], 2))

print("\nHighest Profit Month:", best_profit_month)
print("Highest Monthly Profit ($):",
      round(monthly_analysis.loc[best_profit_month, "Total_Profit"], 2))

print("\nLowest Profit Month:", worst_profit_month)
print("Lowest Monthly Profit ($):",
      round(monthly_analysis.loc[worst_profit_month, "Total_Profit"], 2))
```

# Business Insights and Recommendations

## 1. Overall Business Performance
- Total sales were **$2,297,200.86**, with a total profit of **$286,397.02**.
- The dataset contains **9,994 transaction records**, 5,009 unique orders, and 793 unique customers.
- Although overall profit was positive, some products and sub-categories generated losses.

## 2. Category Performance
- **Technology** generated the highest sales ($836,154.03) and the highest profit ($145,454.95).
- Office Supplies sold the highest quantity, with 22,906 units.
- Furniture generated $741,999.80 in sales but only $18,451.27 in profit, indicating a need to investigate its profitability.

## 3. Monthly Sales and Profit Trends
- November 2017 recorded the highest monthly sales of **$118,447.82**.
- February 2014 recorded the lowest monthly sales of **$4,519.89**.
- December 2016 recorded the highest monthly profit of **$17,885.31**.
- January 2015 recorded the lowest monthly profit, resulting in a loss of **$3,281.01**.

## 4. Regional Performance
- The West region had the highest sales ($725,457.82) and profit ($108,418.45).
- The East region ranked second in sales and profit.
- The Central region generated higher sales than the South but lower profit, suggesting that its profitability deserves further investigation.

## 5. Product and Sub-Category Performance
- The Canon imageCLASS 2200 Advanced Copier was the highest-selling product by revenue, generating $61,599.82 in sales.
- The Tables sub-category recorded a total loss of $17,725.48.
- Some high-revenue products also generated losses, showing that high sales do not always mean high profitability.

## 6. Discount Analysis
- Discount bands above 20% recorded negative total profit in this dataset.
- The 31–50% discount band recorded a total loss of $48,447.73.
- The above-50% discount band recorded a total loss of $76,559.05.
- These patterns suggest that the business should review its discount strategy. However, further analysis is needed to determine whether discounts directly contribute to losses.

## 7. Business Recommendations
1. Investigate Furniture and Tables to identify high costs, excessive discounts, or low-margin products.
2. Review products that generate high sales but low or negative profit.
3. Evaluate the discount policy, especially for discounts above 20%.
4. Study the Central region's costs and product mix to identify opportunities to improve profitability.
5. Monitor monthly sales and profit trends to support inventory planning and business decisions.
6. Continue investing in strong-performing categories and regions while investigating loss-making areas.

## Conclusion
The analysis shows that the business generates substantial revenue and positive overall profit, but profitability varies across products, categories, regions, months, and discount levels. Using these insights can help the business improve pricing decisions, control losses, and make more informed sales strategies.
